````python
def convert(obj):
    L = []
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L
````

````python
def convert2(obj):
    L = []
    counter = 0

    for i in ast.literal_eval(obj):
        if counter <= 3:
            L.append(i['name'])
            counter += 1
        else:
            break
    return L
````

````python
movies['cast'] = movies['cast'].fillna('[]')
````

````python
movies['cast'] = movies['cast'].apply(convert2)
````

# Code Explanation: Remove Spaces from Strings in Multiple Columns

## Code Breakdown

```python
movies['genres'] = movies['genres'].apply(lambda x: [i.replace(" ", "") for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x: [i.replace(" ", "") for i in x])
movies['cast'] = movies['cast'].apply(lambda x: [i.replace(" ", "") for i in x])
movies['crew'] = movies['crew'].apply(lambda x: [i.replace(" ", "") for i in x])
```

## Components

### 1. **`movies['column_name']`**
- Selects a specific column from the pandas DataFrame
- Applied to: `genres`, `keywords`, `cast`, `crew`

### 2. **`.apply()`**
- A pandas method that applies a function to each element in the Series
- Executes the function row-by-row

### 3. **`lambda x:`**
- Anonymous function that takes one parameter `x`
- `x` represents each individual element (a list) in the column

### 4. **`[i.replace(" ","") for i in x]`**
- **List comprehension** that iterates through each item `i` in the list `x`
- **`.replace(" ","")`** method removes all spaces from each string
  - First parameter: `" "` (space to find)
  - Second parameter: `""` (replace with nothing)
- Returns a new list with spaces removed

## Example

**Input:**
```python
['Science Fiction', 'Action Adventure']
```

**Output:**
```python
['ScienceFiction', 'ActionAdventure']
```

## Why This Matters

Removing spaces from genre, keyword, cast, and crew names:
- Ensures consistency in data representation
- Prevents issues when creating combined tags
- Makes text processing more uniform for recommendation algorithms
- Helps with string matching and comparison

````python
movies['tags']  = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
````

````python
new_def = movies[['movie_id', 'title', 'tags']]
````

````python
new_def.duplicated().sum()
````

````python
new_def['tags'].apply(lambda x : " ".join(x))
````

````python
new_def['tags'] = new_def['tags'].apply(lambda x: " ".join(x).lower())
````

````python
import nltk
````

````python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
````

````python
def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
````

````python
new_def['tags'] = new_def['tags'].apply(stem)
````

````python
new_def
````

````python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000, stop_words= 'english')
````

````python
vectors = cv.fit_transform(new_def['tags']).toarray()
````

````python
vectors
````

````python
cv.get_feature_names_out()
````

````python
ps.stem('lovely')
````

````python
from sklearn.metrics.pairwise import cosine_similarity
````

````python
similarity = cosine_similarity(vectors)
````

````python
sorted(list(enumerate(similarity[0])),reverse=True,key=lambda x:x[1])[1:6]
````

````python
def recommend(movie):
    movies_index = new_def[new_def['title'] == movie].index[0]
    distances = similarity[movies_index]  
    movies_list = sorted(list(enumerate(distances)),reverse=True,key=lambda x:x[1])[1:6]
    for i in movies_list:
        print(new_def.iloc[i[0]].title)
             
````

````python
recommend('Batman Begins')
````
